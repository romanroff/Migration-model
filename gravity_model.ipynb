```python
import skmob
from skmob.utils import utils, constants
from skmob.models.gravity import Gravity

import numpy as np
import pandas as pd
import geopandas as gpd
```

```python
url_tess = skmob.utils.constants.NY_COUNTIES_2011
tessellation = gpd.read_file(url_tess).rename(columns={'tile_id': 'tile_ID'})
```

```python
fdf = skmob.FlowDataFrame.from_file(skmob.utils.constants.NY_FLOWS_2011,
                                        tessellation=tessellation,
                                        tile_id='tile_ID',
                                        sep=",")
```

```python
tot_outflows = fdf[fdf['origin'] != fdf['destination']].groupby(by='origin', axis=0)[['flow']].sum().fillna(0) # Группировка по городам и подсчет миграции из каждого города
tessellation = tessellation.merge(tot_outflows, left_on='tile_ID', right_on='origin').rename(columns={'flow': constants.TOT_OUTFLOW}) # Merge с городами, населением и геометрией
```

```python
gravity_singly = Gravity(gravity_type='singly constrained', origin_exp=0.5, destination_exp=2.0, deterrence_func_args=[-4.0])
print(gravity_singly)
```

```python
gravity_singly.fit(fdf, relevance_column='population')
print(gravity_singly)
```

```python
np.random.seed(0)

synth_fdf = gravity_singly.generate(tessellation,
                                   tile_id_column='tile_ID',
                                   tot_outflows_column='tot_outflow',
                                   relevance_column= 'population',
                                   out_format='flows')

synth_fdf
```

```python
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut

# Инициализация геокодера
geolocator = Nominatim(user_agent="geoapi")

# Функция для получения координат по названию города с обработкой ошибок
def get_coordinates(city_name):
    try:
        location = geolocator.geocode(city_name, timeout=10)  # Устанавливаем таймаут 10 секунд
        if location:
            print(f"City: {city_name}, location: {location.longitude, location.latitude}")
            return (location.longitude, location.latitude)
        else:
            return (None, None)
    except GeocoderTimedOut:
        print(f"TimeoutError: Не удалось получить координаты для города {city_name}")
        return (None, None)
    except Exception as e:
        print(f"Ошибка: {e} при получении координат для города {city_name}")
        return (None, None)

# Пример использования
city_names = NY['origin'].unique()
coordinates = {city: get_coordinates(city) for city in city_names}
```

```python
import geopandas as gpd
from shapely.geometry import Point

NY['coordinates'] = NY['origin'].map(coordinates)

# Разделяем координаты на отдельные столбцы longitude и latitude
NY[['longitude', 'latitude']] = pd.DataFrame(NY['coordinates'].tolist(), index=NY.index)
# Удаляем временный столбец coordinates
NY.drop(columns=['coordinates'], inplace=True)

# Создаем столбец geometry на основе координат
NY['geometry'] = NY.apply(lambda row: Point(row['longitude'], row['latitude']), axis=1)
NY.drop(columns=['latitude', 'longitude'], inplace=True)

# Преобразуем DataFrame в GeoDataFrame
gdf = gpd.GeoDataFrame(NY, geometry='geometry')

# Устанавливаем систему координат (например, WGS84)
gdf.set_crs(epsg=4326, inplace=True)
gdf.drop(columns=['destination', 'population_d', 'flow'], inplace=True)
gdf.drop_duplicates(inplace=True)
gdf = gdf.rename(columns={'population_o': 'population', 'origin': 'tile_ID'})

# Результат
gdf
```

```python
import geopandas as gpd
import pandas as pd

# Объединяем данные для origin
synth_fdf2 = pd.merge(
    synth_fdf,
    synth_fdf.tessellation[['tile_ID', 'geometry']],
    left_on='origin',
    right_on='tile_ID',
    how='left'
)
synth_fdf2 = synth_fdf2.rename(columns={'geometry': 'origin_geometry'})

# Объединяем данные для destination
synth_fdf2 = pd.merge(
    pd.DataFrame(synth_fdf2),
    synth_fdf.tessellation[['tile_ID', 'geometry']],
    left_on='destination',
    right_on='tile_ID',
    how='left'
)
synth_fdf2 = synth_fdf2.rename(columns={'geometry': 'destination_geometry'})

# Удаляем лишние столбцы
synth_fdf2 = synth_fdf2.drop(columns=['tile_ID_x', 'tile_ID_y'])
```

```python
top_quartile_threshold = synth_fdf2['flow'].quantile(0.1)                 # Вычисляем 75-й процентиль для flow
top_quartile_df = synth_fdf2[synth_fdf2['flow'] >= top_quartile_threshold]  # Фильтруем данные, оставляя только значения выше 75-го процентиля
top_quartile_df = top_quartile_df[top_quartile_df['origin'] != top_quartile_df['destination']]
# top_quartile_df['flow'] = top_quartile_df['flow'] / top_quartile_df['flow'].max()
```

```python
import folium
import branca.colormap as cm

# Создаем базовую карту
m = folium.Map(location=[40.7128, -74.0060], zoom_start=10)  # Центр карты (например, Нью-Йорк)

# Создаем линейную цветовую шкалу
colormap = cm.LinearColormap(
    colors=['blue', 'green', 'yellow', 'red'],
    vmin=0,  # Минимальное значение flow
    vmax=top_quartile_df['flow'].max()   # Максимальное значение flow
)

# Добавляем шкалу на карту для легенды
colormap.caption = 'Flow Gradient'
m.add_child(colormap)

# Добавляем линии с градиентными цветами
for _, row in top_quartile_df.iterrows():
    origin = row['origin_geometry']
    destination = row['destination_geometry']
    flow = row['flow']
    
    # Координаты для линии
    line_coords = [(origin.centroid.y, origin.centroid.x), (destination.centroid.y, destination.centroid.x)]
    
    # Цвет линии зависит от значения flow
    line_color = colormap(flow)
    
    # Добавляем линию на карту
    folium.PolyLine(
        locations=line_coords,
        color=line_color,
        weight=flow / 10000,  # Толщина линии зависит от flow
        opacity=0.8,
        tooltip=f"Flow: {flow}"
    ).add_to(m)

# Добавляем маркеры для городов
for _, row in synth_fdf.tessellation.iterrows():
    folium.Marker(
        location=[row['geometry'].centroid.y, row['geometry'].centroid.x],
        popup=row['tile_ID'],
        icon=folium.Icon(color='red', icon='info-sign')
    ).add_to(m)

# Отображаем карту в Jupyter Notebook
m
```
